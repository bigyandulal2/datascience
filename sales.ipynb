```python
import pandas as pd
df=pd.read_csv("Sales-Practice.csv");
```

```python
df.head()
```

```python
#total amount sold per category
amt_per_category=df.groupby("Category").agg(
    total_sales=("Amount","sum")

)
print(amt_per_category)
```

```python
#average amount per city
avg_per_city=df.groupby("City").agg(
    avg_amount=("Amount","mean")
)
print(avg_per_city)
```

```python
#orders per city
avg_order_per_city=df.groupby("City").agg(
    avg_order=("Quantity","sum")
)
print(avg_order_per_city)
```

```python
#total quantity per city and category together
total_quantity_per_city_category=df.groupby(["City","Category"]).agg(
    total_quantity=("Quantity","sum"),
      quantity_count=("Quantity", "count")

)
print(total_quantity_per_city_category);
```
