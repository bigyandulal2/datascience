```python
import pandas as pd
df=pd.read_csv("Students-Practice.csv")
from google.colab import drive
drive.mount('/content/drive')
df.to_csv("/content/drive/MyDrive/myfile.csv", index=False)
```

```python
df.head()
```

```python
# Average Marks per Class? (A vs B — which class is better?)
avg_marks=df.groupby("Class").agg(
    Marks=("Marks","mean")
)
print(avg_marks)
```

```python
# Average Marks per Subject? (Math vs Science)
avg_marks_sub=df.groupby("Subject").agg(
    Marks=("Marks","mean")
)
print(avg_marks_sub)
```

```python
# How many students per Gender? (use count)
count_gender=df.groupby("Gender").agg(
    Count=("Gender","count")
)
print(count_gender)
```

```python
# Average Marks per Class and Gender? Boy A or girl B
avg_marks_class_gender=df.groupby(["Class","Gender"]).agg(
    Marks=("Marks","mean")
)
print(avg_marks_class_gender)
```

```python
# Maximum Marks per Subject?
max_marks_per_subject=df.groupby("Subject")["Marks"].max()
print(max_marks_per_subject)
```
